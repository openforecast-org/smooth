````python
import os, subprocess, sys
ROOT = os.path.abspath(os.path.join(os.getcwd(), '..', '..', '..'))
CODE = os.path.join(ROOT, 'claimset-code', 'msarima-interval-calibration')
RUNS = os.path.join(ROOT, 'claimset', 'msarima-interval-calibration', 'lab', 'runs')
NB = os.path.join(ROOT, '..', 'smooth', 'python', 'tests', 'notebooks')
env = dict(os.environ, CLAIMSET_STUDY='msarima-interval-calibration', OMP_NUM_THREADS='1', OPENBLAS_NUM_THREADS='1')
env.pop('CLAIMSET_RESULTS', None)

def sh(*args):
    r = subprocess.run(list(args), cwd=ROOT, env=env, capture_output=True, text=True)
    print((r.stdout + r.stderr)[-6000:])
````
